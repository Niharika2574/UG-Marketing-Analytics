```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel(r"C:\Users\nihar\Downloads\RFM for Python.xlsx",usecols=["Customer","Date","Amount"])
df
```

```python
today=df["Date"].max()
today
```

```python
#Build R, F, M
rfm=df.groupby("Customer").agg(
    R=("Date", lambda x: (today-x.max()).days),
    F=("Date","count"),
    M=("Amount","sum"),
)
rfm
```

```python
#Score 1-5 by quintile
rfm["R_score"]=pd.qcut(rfm["R"],5,labels=[5,4,3,2,1]).astype(int)
rfm["F_score"]=pd.qcut(rfm["R"],5,labels=[1,2,3,4,5]).astype(int)
rfm["M_score"]=pd.qcut(rfm["R"],5,labels=[1,2,3,4,5]).astype(int)
rfm
```

```python
rfm["RFM"]=(rfm["R_score"].astype(str)+rfm["F_score"].astype(str)+rfm["M_score"].astype(str))
```

```python
rfm["Score"]=rfm["R_score"]+rfm["F_score"]+rfm["M_score"]
```
