```python
import pandas as pd
 
df = pd.read_excel("/Users/sunitakumar/Desktop/New_data/marketing_analytics/RFM _practice.xlsx")
today = df["Date"].max()
 
rfm = df.groupby("Customer").agg(
    R=("Date", lambda x: (today - x.max()).days),
    F=("Date", "count"),
    M=("Amount", "sum"),
)
 
rfm["R_score"] = pd.qcut(rfm["R"], 5, labels=[5, 4, 3, 2, 1])
rfm["F_score"] = pd.qcut(rfm["F"], 5, labels=[1, 2, 3, 4, 5])
rfm["M_score"] = pd.qcut(rfm["M"], 5, labels=[1, 2, 3, 4, 5])
 
rfm.to_excel("rfm.xlsx")
print(rfm.head())
```
