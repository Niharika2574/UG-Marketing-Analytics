```python
import pandas as pd
df=pd.read_excel(r"C:\Users\nihar\Downloads\RFM for Python.xlsx")
df
```

```python
today=df["Date"].max()
today
```

```python
rfm=df.groupby("Customer").agg(
    R=("Date",lambda x: (today-x.max()).days),
    F=("Date","count"),
    M=("Amount","sum"),
)
```

```python
rfm
```

```python
rfm["R_score"]=pd.qcut(rfm["R"],5,labels=[5,4,3,2,1])
rfm["F_score"]=pd.qcut(rfm["R"],5,labels=[1,2,3,4,5])
rfm["M_score"]=pd.qcut(rfm["R"],5,labels=[1,2,3,4,5])
```

```python
rfm.to_excel("RFM for Python.xlsx")
print(rfm.head())
```
